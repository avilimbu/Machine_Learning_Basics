```python
import numpy as np
```

```python
# arange | returns a 1D NumPy array (ndarray) containing evenly spaced values within a specified interval.
# arange(start,end,step)
a = np.arange(1,10,2)
a
```

```python
a = np.arange(1,20,2.7)
a
```

The np.linspace() function in Python's NumPy library returns an array of evenly (linearly) spaced numbers over a specified interval. Instead of choosing the step size, you tell it how many numbers you want, and NumPy automatically calculates the exact spacing.

```python
# linspace | linspace(start, end, no.of elements)
b = np.linspace(1,10,6)
b
```

```python
# arange | reshape()
a = np.arange(10).reshape(2,5)
a
```

```python
# 3D arary
a = np.arange(24).reshape(2,3,4)
a
```
